```python
print("I am your AI Travel Agent!")
print("-" * 55)
print("Please tell me your destination.")

place = input("Enter destination: ").strip().lower()

if place == "dubai":
    price = 150000
    available = "Yes"
else:
    if place == "turkey":
        price = 180000
        available = "Yes"
    else:
        if place == "malaysia":
            price = 130000
            available = "Yes"
        else:
            if place == "saudi arabia":
                price = 110000
                available = "Yes"
            else:
               if place == "saudi arabia":
                  price = 110000
                  available = "Yes"
               else:
                available = "No"

if available == "Yes":
    print("Tickets are available!")
    print("Ticket price:", price, "Rs")

    budget = int(input("Enter your budget in Rs: "))

    if budget >= price:
        confirm = input("Do you want to book this ticket? (yes/no): ")

        if confirm == "yes":
            print("Your ticket has been booked successfully!")
        else:
            print("Booking cancelled.")
    else:
        print("Sorry, your budget is too low.")
else:
    print("Sorry, tickets are not available for this destination.")
```
